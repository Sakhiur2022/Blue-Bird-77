```python
import tensorflow as tf
import numpy as np
import os

# Download Fashion-MNIST
(train_images, train_labels), (test_images, test_labels) = (
    tf.keras.datasets.fashion_mnist.load_data()
)

print("Train images:", train_images.shape)
print("Train labels:", train_labels.shape)

print("Test images: ", test_images.shape)
print("Test labels: ", test_labels.shape)
```

# 2 - Save it into  fmnist/ folder

```python
FMNIST_DIR = "/content/fmnist"

os.makedirs(FMNIST_DIR, exist_ok=True)

np.save(
    os.path.join(FMNIST_DIR, "train_images.npy"),
    train_images
)

np.save(
    os.path.join(FMNIST_DIR, "train_labels.npy"),
    train_labels
)

np.save(
    os.path.join(FMNIST_DIR, "test_images.npy"),
    test_images
)

np.save(
    os.path.join(FMNIST_DIR, "test_labels.npy"),
    test_labels
)

print("Saved Fashion-MNIST to:", FMNIST_DIR)
```

# 7 - Multi-image interactive inspector

```python
import random
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output


# Widgets


split_dropdown = widgets.Dropdown(
    options=["Train", "Test"],
    value="Train",
    description="Split:"
)

class_dropdown = widgets.Dropdown(
    options=["All"] + CLASS_NAMES,
    value="All",
    description="Class:"
)

num_slider = widgets.IntSlider(
    value=12,
    min=1,
    max=36,
    step=1,
    description="Images:",
    continuous_update=False
)

seed_slider = widgets.IntSlider(
    value=42,
    min=0,
    max=9999,
    step=1,
    description="Seed:",
    continuous_update=False
)

show_button = widgets.Button(
    description="Show Images",
    button_style="primary"
)

output = widgets.Output()



# Display images


def show_images(_=None):

    with output:
        clear_output(wait=True)

        if split_dropdown.value == "Train":
            images = train_images
            labels = train_labels
        else:
            images = test_images
            labels = test_labels

        selected_class = class_dropdown.value

        # Select candidate indices
        if selected_class == "All":

            candidate_indices = list(range(len(images)))

        else:

            class_id = CLASS_NAMES.index(selected_class)

            candidate_indices = np.where(
                labels == class_id
            )[0].tolist()

        # Reproducible selection
        random.seed(seed_slider.value)

        num_images = min(
            num_slider.value,
            len(candidate_indices)
        )

        selected_indices = random.sample(
            candidate_indices,
            num_images
        )

        # Grid
        cols = 6
        rows = (num_images + cols - 1) // cols

        fig, axes = plt.subplots(
            rows,
            cols,
            figsize=(15, 2.8 * rows)
        )

        axes = np.array(axes).reshape(-1)

        # Draw images
        for ax, idx in zip(axes, selected_indices):

            ax.imshow(
                images[idx],
                cmap="gray"
            )

            ax.set_title(
                f"{CLASS_NAMES[labels[idx]]}\n"
                f"index={idx}",
                fontsize=9
            )

            ax.axis("off")

        # Hide unused axes
        for ax in axes[num_images:]:
            ax.axis("off")

        plt.tight_layout()
        plt.show()

        print(f"Split: {split_dropdown.value}")
        print(f"Class: {selected_class}")
        print(f"Images displayed: {num_images}")
        print(f"Random seed: {seed_slider.value}")


show_button.on_click(show_images)

# Display inspector


display(
    widgets.VBox([
        widgets.HTML(
            "<h3>Fashion-MNIST Inspector</h3>"
        ),

        split_dropdown,
        class_dropdown,
        num_slider,
        seed_slider,

        show_button,

        output
    ])
)

show_images()
```

# 8 - Individual interactive inspector

```python
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# -----------------------------------
# Widgets
# -----------------------------------

split_widget = widgets.Dropdown(
    options=["Train", "Test"],
    value="Train",
    description="Split:"
)

index_widget = widgets.IntSlider(
    value=0,
    min=0,
    max=len(train_images) - 1,
    step=1,
    description="Index:",
    continuous_update=False
)

previous_button = widgets.Button(
    description="← Previous"
)

next_button = widgets.Button(
    description="Next →"
)

random_button = widgets.Button(
    description="🎲 Random"
)

output = widgets.Output()


# Get selected dataset


def get_dataset():

    if split_widget.value == "Train":
        return train_images, train_labels

    return test_images, test_labels



# Display selected image


def show_sample():

    with output:
        clear_output(wait=True)

        images, labels = get_dataset()

        idx = index_widget.value

        image = images[idx]
        label = labels[idx]

        plt.figure(figsize=(6, 6))

        plt.imshow(
            image,
            cmap="gray",
            vmin=0,
            vmax=255
        )

        plt.title(
            f"{split_widget.value}\n"
            f"Index {idx} — {CLASS_NAMES[label]}",
            fontsize=14
        )

        plt.axis("off")
        plt.show()

        print("────────────────────────────")
        print("Fashion-MNIST Sample")
        print("────────────────────────────")
        print(f"Split:       {split_widget.value}")
        print(f"Index:       {idx}")
        print(f"Label ID:    {label}")
        print(f"Class:       {CLASS_NAMES[label]}")
        print(f"Image shape: {image.shape}")
        print(f"Data type:   {image.dtype}")
        print(
            f"Pixel range: {image.min()} - {image.max()}"
        )
        print("────────────────────────────")


# Previous


def previous_sample(_):

    if index_widget.value > 0:
        index_widget.value -= 1

    show_sample()



# Next


def next_sample(_):

    if index_widget.value < index_widget.max:
        index_widget.value += 1

    show_sample()



# Random


def random_sample(_):

    import random

    index_widget.value = random.randint(
        0,
        index_widget.max
    )

    show_sample()


# Split changed


def split_changed(change):

    if change["new"] == "Train":

        index_widget.max = len(train_images) - 1

    else:

        index_widget.max = len(test_images) - 1

    index_widget.value = 0

    show_sample()



# Connect widgets


previous_button.on_click(previous_sample)

next_button.on_click(next_sample)

random_button.on_click(random_sample)

split_widget.observe(
    split_changed,
    names="value"
)


# Display


display(
    widgets.VBox([

        widgets.HTML(
            "<h3>Individual Fashion-MNIST Inspector</h3>"
        ),

        split_widget,

        index_widget,

        widgets.HBox([
            previous_button,
            next_button,
            random_button
        ]),

        output
    ])
)

show_sample()
```
